```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('practice.db')

def run_query(sql):
    return pd.read_sql_query(sql, conn)
```

```python
run_query("SELECT name, department, salary FROM employees WHERE department = 'Engineering'")
```

```python
run_query("SELECT name, salary FROM employees WHERE salary > 50000")
```

```python
run_query("SELECT name, department, hire_year FROM employees WHERE department = 'Sales' AND hire_year > 2020")
```

```python
run_query("SELECT name, department FROM employees WHERE department = 'Sales' OR department = 'Marketing'")
```

```python
run_query("SELECT name, department FROM employees WHERE department IN ('Sales', 'Marketing')")
```

```python
run_query("SELECT name, department, salary FROM employees ORDER BY department, salary DESC")
```

```python
run_query("SELECT name, department, salary FROM employees WHERE department = 'Engineering' ORDER BY salary DESC")
```

```python
run_query("SELECT department, COUNT(*) AS num_employees FROM employees GROUP BY department")
```

```python
run_query("SELECT department, AVG(salary) AS avg_salary FROM employees GROUP BY department")
```

```python
run_query("SELECT department,   COUNT(*) AS num_employees, SUM(salary) AS total_salary,AVG(salary) AS avg_salary,MAX(salary) AS highest_salary,MIN(salary) AS lowest_salary FROM employees GROUP BY department")
```

```python
run_query("SELECT department, AVG(salary) AS avg_salary FROM employees GROUP BY department HAVING AVG(salary) > 50000")
```

```python
run_query("SELECT orders.order_id, employees.name, orders.product, orders.amount FROM orders INNER JOIN employees ON orders.emp_id = employees.emp_id")
```

```python
run_query("SELECT e.name, o.order_id, o.product, o.amount FROM employees e LEFT JOIN orders o ON e.emp_id = o.emp_id")
```

```python
run_query("SELECT e.name, SUM(o.amount) AS total_orders FROM employees e LEFT JOIN orders o ON e.emp_id = o.emp_id GROUP BY e.name ORDER BY total_orders DESC")
```
